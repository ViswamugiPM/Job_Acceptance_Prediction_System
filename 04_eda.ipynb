```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/processed/cleaned_job_acceptance_data.csv")
```

```python
df.head()
df.shape
```

```python
df['status'].value_counts()
```

```python
sns.countplot(x='status', data=df)
plt.title("Job Placement Status Distribution")
plt.show()
```

```python
plt.hist(df['age_years'], bins=20)
plt.title("Age Distribution of Candidates")
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
plt.hist(df['skills_match_percentage'], bins=20)
plt.title("Skills Match Percentage Distribution")
plt.show()
```

```python
sns.boxplot(x='status', y='technical_score', data=df)
plt.title("Technical Score vs Placement Status")
plt.show()
```

```python
sns.boxplot(x='status', y='skills_match_percentage', data=df)
plt.title("Skills Match vs Placement Status")
plt.show()
```

```python
sns.countplot(x='company_tier', hue='status', data=df)
plt.title("Company Tier vs Placement Status")
plt.show()
```

```python
sns.boxplot(x='status', y='years_of_experience', data=df)
plt.title("Experience vs Placement Status")
plt.show()
```

```python
sns.countplot(x='competition_level', hue='status', data=df)
plt.title("Competition Level vs Placement Status")
plt.show()
```

CORRELATION ANALYSIS


```python
plt.figure(figsize=(12,8))
sns.heatmap(df.select_dtypes(include=np.number).corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
